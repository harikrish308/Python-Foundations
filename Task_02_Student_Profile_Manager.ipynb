```python
# Task 2 - Student Profile Manager
name = input("Name: ")
age = int(input("Age: "))
department = input("Department: ");
cgpa = float(input("CGPA: "))
blood_group = input("Blood Group: ")
email = input("Email: ")
phone = input("Phone: ")

print("\n" + "=" * 40)
print("         STUDENT PROFILE")
print("=" * 40)
print(f"Name        : {name}")
print(f"Age         : {age}")
print(f"Department  : {department}")
print(f"CGPA        : {cgpa}")
print(f"Blood Group : {blood_group}")
print(f"Email       : {email}")
print(f"Phone       : {phone}")
print("=" * 40)
```
